## Analise de um dataset de compras de um e - commerce

```python
import pandas as pd

# Criar um dataframe
df = pd.read_csv('Ecommerce Purchases.csv')

# Ver apenas as primeiras linhas
display(df.head())
```

```python
# Imprimi as colunas
# print(df.columns)

for i in df.columns:
    print(i)
```

```python
# Ver apenas uma coluna
print(df['Language'].head(2))

print('\n------------------------------------------\n')

# Ver mais que uma coluna
print(df[['Company','Language','CC Provider']].head(2))
```

```python
# Agrupar as vendas por idioma
idioma = df.groupby('Language')['Language'].count()
print(idioma)

job = df.groupby('Job')['Job'].count()
print(job)
```

```python
# Somar uma coluna
total = df["Purchase Price"].sum()

print(f'Faturamento de vendas : $ {total:.2f} dólares')

# Vendas por tipo de cartão usado
vendasCartao = df.groupby('CC Provider')['Purchase Price' ].sum()
print(vendasCartao)

print('\n-----------------------------------------\n')

# Apresente a média de vendas por idioma (mean)
mediaCartao = df.groupby('Language')['Purchase Price'].mean()
print(mediaCartao)
```

```python
# Criar filtros

# Ver apenas as vendas de um idioma especifico

idioma_el = df[df['Language'] =='el']
display(idioma_el[['Job','Language']].head(10))
```

```python
# Criar duas colunas
# Coluna 1 - Imposto de 10% sobre as vendas 
df['imposto'] = df['Purchase Price'] * 0.10

display(df[['Purchase Price','imposto']])
# Coluna 2 - Calcular o valor da venda - o imposto 
# ValorSemImposto

df['ValorSemImposto'] = df['Purchase Price'] - df['imposto'] 

display(df[['Purchase Price','imposto','ValorSemImposto']])
```
